```python
def inspeccionar_autocorrelacion(R, f0, fs, t, tiempos,
                                 f0_max=600, threshold=0.6):
    """
    Muestra la autocorrelación para los instantes indicados.

    Parameters
    ----------
    R : ndarray
        Matriz de autocorrelación (num_frames x max_lag).
    f0 : ndarray
        Frecuencia fundamental estimada por frame.
    fs : int
        Frecuencia de muestreo.
    t : ndarray
        Tiempo asociado a cada frame.
    tiempos : list
        Instantes (en segundos) a inspeccionar.
    """

    lag_min = int(fs / f0_max)

    for t_n in tiempos:

        # Frame cuyo centro está más cerca de t_n
        n = np.argmin(np.abs(t - t_n))

        R_norm = R[n] / np.max(R[n])

        plt.figure(figsize=(10,5))
        plt.title(f'Autocorrelación en t = {t[n]:.2f} s')

        plt.plot(R_norm, label='Autocorrelación')
        plt.axhline(threshold, ls='--', c='g',
                    label='Umbral')
        plt.axvline(lag_min, ls='--', c='k',
                    label=f'Lag mínimo ({f0_max} Hz)')

        if f0[n] > 0:
            lag = int(round(fs / f0[n]))
            plt.plot(lag, R_norm[lag], '*r',
                     ms=12, label=f'f0 = {f0[n]:.1f} Hz')

        plt.xlabel('Retardo (muestras)')
        plt.ylabel('Autocorrelación normalizada')
        plt.xlim(0, R.shape[1]-1)
        plt.grid(True)
        plt.legend()
        plt.show()

def mostrar_f0_espectrograma(x, fs, t, f0, gt_file=None):

    # Espectrograma
    f, tt, Sxx = signal.spectrogram(
        x,
        fs=fs,
        window='hann',
        nperseg=int(0.03*fs),
        noverlap=int(0.02*fs),
        scaling='spectrum'
    )

    plt.figure(figsize=(12,6))

    plt.pcolormesh(
        tt,
        f,
        10*np.log10(Sxx + 1e-12),
        shading='gouraud',
        cmap='viridis'
    )

    # Estimación
    plt.plot(t, f0, 'r', lw=2, label='Estimación')

    # Ground truth
    if gt_file is not None:

        gt = np.loadtxt(gt_file)

        t_gt = gt[:,0]
        f0_gt = gt[:,1]

        # Interpolar sobre los tiempos de la estimación
        f0_gt_interp = np.interp(t, t_gt, f0_gt, left=0, right=0)

        plt.plot(t,f0_gt_interp,'--w', lw=2, label='Ground truth')

    plt.xlabel('Tiempo [s]')
    plt.ylabel('Frecuencia [Hz]')
    plt.ylim(0,1000)

    plt.colorbar(label='Potencia [dB]')
    plt.legend()
    plt.tight_layout()
    plt.show()
```
