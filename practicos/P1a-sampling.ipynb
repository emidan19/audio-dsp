1. Abrir el archivo *tones.wav* (muestreado a $44100Hz$), que contiene un tono de $6000Hz$ seguido de otro tono de $2000Hz$. Se quiere re-muestrear esta señal a $22050Hz$ y a $11025Hz$. Realizarlo de dos formas: 

    a) descartando una de cada dos muestras 

    b) utilizando la función *decimate* de la biblioteca [scipy.signal](https://docs.scipy.org/doc/scipy/reference/signal.html).  

Comparar los resultados, escuchando los archivos de audio obtenidos y mediante visualzación de espectrogramas. Se recomienda utilizar la función *specgram* de *matplotlib* o *spectrogram* de *scipy*.

```python
sr = 44100
N = 2 * sr
t = np.linspace(0, N / sr, N)

# Tiempos y frecuencia tono 6kHz
t_6k = t.copy()
t_6k[N//2:] = 0
f_6k = 6000
# Tiempos y frecuencia tono 2kHz
t_2k = t.copy()
t_2k[:N//2] = 0
f_2k = 2000

# Tonos
tones = np.cos(2 * np.pi * (f_6k * t_6k + f_2k * t_2k))

print("¡Cuidado con el volumen del audio! Se recomienda bajar el volumen")
display(Audio(tones, rate=sr))

f, t, Sxx = signal.spectrogram(tones,sr)
plt.pcolormesh(t, f, Sxx, shading='gouraud')
plt.title('Tonos originales a 44100 kHz')
plt.ylabel('Frequency [Hz]')
plt.xlabel('Time [sec]')
plt.show()

```
